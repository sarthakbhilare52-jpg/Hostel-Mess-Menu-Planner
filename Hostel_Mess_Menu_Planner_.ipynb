```python

import random

# 1. Database of available dishes
dishes = {
    "Breakfast": [
        "Poha", "Upma", "Idli", "Dosa", "Paratha",
        "Sandwich", "Puri Bhaji", "Bread Butter",
        "Misal Pav", "Cornflakes"
    ],

    "Lunch": [
        "Dal Rice", "Paneer Masala", "Veg Pulao",
        "Chole Bhature", "Rajma Rice", "Mix Veg",
        "Aloo Gobi", "Veg Biryani", "Dal Khichdi",
        "Pav Bhaji"
    ],

    "Dinner": [
        "Chapati Bhaji", "Jeera Rice", "Palak Paneer",
        "Veg Fried Rice", "Dal Tadka", "Pav Usal",
        "Matar Paneer", "Veg Kofta", "Masala Dosa",
        "Kadhi Rice"
    ]
}

# 2. Premium dishes for Sunday
premium_dishes = [
    "Special Thali",
    "Paneer Tikka",
    "Veg Biryani",
    "Pav Bhaji",
    "Shahi Paneer"
]

# 3. Create voting dictionary
votes = {}

for meal in dishes:
    for dish in dishes[meal]:
        votes[dish] = 0

# 4. Create feedback list
feedback_list = []

# 5. Student voting and feedback system
print("===== HOSTEL MESS MENU PLANNER =====")

while True:

    print("\n1. Vote for favourite dish")
    print("2. Add Feedback")
    print("3. Finish and Generate Menu")

    choice = input("\nEnter your choice (1-3): ").strip()

    if choice == "1":

        print("\n===== AVAILABLE DISHES =====")

        for dish in votes:
            print("-", dish)

        selected = input(
            "\nEnter your favourite dish: "
        ).strip().lower()

        matched_dish = None

        for dish in votes:
            if dish.lower() == selected:
                matched_dish = dish
                break

        if matched_dish is not None:
            votes[matched_dish] += 1
            print("Your vote has been recorded!")

        else:
            print("Invalid dish! Please try again.")

    elif choice == "2":

        print("\n===== ADD YOUR FEEDBACK =====")

        feedback = input(
            "Enter your feedback about hostel food: "
        ).strip()

        if feedback:
            feedback_list.append(feedback)
            print("Thank you! Your feedback has been saved.")

        else:
            print("Feedback cannot be empty.")

    elif choice == "3":
        print("\nGenerating weekly menu...")
        break

    else:
        print("Invalid choice! Please enter 1, 2 or 3.")

# 6. Function to select dishes using votes
def select_dish(available_dishes):

    weights = []

    for dish in available_dishes:
        weights.append(votes[dish] + 1)

    selected = random.choices(
        available_dishes,
        weights=weights,
        k=1
    )[0]

    available_dishes.remove(selected)

    return selected

# 7. Generate 7-day weekly menu
days = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

weekly_menu = {}

for day in days:

    weekly_menu[day] = {}

    for meal in dishes:

        if day == "Sunday" and meal == "Dinner":
            continue

        available_dishes = dishes[meal].copy()

        # Remove dishes already selected in this meal
        for previous_day in weekly_menu:
            if meal in weekly_menu[previous_day]:
                if weekly_menu[previous_day][meal] in available_dishes:
                    available_dishes.remove(
                        weekly_menu[previous_day][meal]
                    )

        weekly_menu[day][meal] = select_dish(
            available_dishes
        )

# 8. Display weekly menu
print("\n")
print("=" * 75)
print("                 HOSTEL MESS WEEKLY MENU")
print("=" * 75)

print(
    f"{'Day':<15}"
    f"{'Breakfast':<20}"
    f"{'Lunch':<20}"
    f"{'Dinner':<20}"
)

print("-" * 75)

for day in days:

    print(
        f"{day:<15}"
        f"{weekly_menu[day]['Breakfast']:<20}"
        f"{weekly_menu[day]['Lunch']:<20}"
        f"{weekly_menu[day].get('Dinner', 'Special Menu'):<20}"
    )

print("=" * 75)

# 9. Special Sunday dish
special_sunday = random.choice(premium_dishes)

print("\n===== SPECIAL SUNDAY DISH =====")
print("Today's special dish is:", special_sunday)

# 10. Display voting results
print("\n===== STUDENT VOTING RESULTS =====")

for dish, count in votes.items():

    if count > 0:
        print(f"{dish:<20} Votes: {count}")

# 11. Display student feedback
print("\n===== STUDENT FEEDBACK =====")

if feedback_list:

    for number, feedback in enumerate(feedback_list, start=1):
        print(f"{number}. {feedback}")

else:
    print("No feedback received.")

print("\nThank you for using Hostel Mess Menu Planner!")
```
